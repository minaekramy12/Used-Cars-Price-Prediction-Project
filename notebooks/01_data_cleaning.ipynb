After Investigating Unique Values

```python
def transmission_mapping(x):

    if pd.isna(x):
        return np.nan

    x = str(x)

    if x in ['Transmission Overdrive Switch',
            '–',
            '2',
            '6-Speed',
            'Single-Speed Fixed Gear',
            '7-Speed',
            'SCHEDULED FOR OR IN PRODUCTION']:
        return np.nan


    elif x in ['Transmission w/Dual Shift Mode']:
        return 'Dual Clutch'


    elif x in ['Variable', 'CVT']:
        return 'CVT'


    elif 'AT/MT' in x:
        return 'Automatic'


    elif any(word in x for word in ['A/T',
                                    'AT',
                                    'Automatic',
                                    'Auto']):
        return 'Automatic'


    elif any(word in x for word in ['M/T',
                                    'MT',
                                    'Manual']):
        return 'Manual'


    return np.nan
```

```python
data.transmission = data.transmission.apply(transmission_mapping)
```
