## Installing Dependencies

```python
import pandas as pd
from scipy import stats
```

```python
data = pd.read_csv('../data/clean_data.csv')
data.head()
```

```python
data.price.isnull().sum()
```

## Does accident history affect car prices?

Null Hypothesis: Cars with accidents have the seame mean price as cars without accidents

Alternate Hypothesis: Cars with accidents have a different mean price

```python
data.accident.unique()
```

```python
with_accident = data[data.accident == 'At least 1 accident or damage reported']['price']
without_accident = data[data.accident != 'At least 1 accident or damage reported']['price']
```

```python
t_stat, p_value = stats.ttest_ind(without_accident, with_accident)
p_value
```

Since P-Value is less than 0.05. So we will reject the null hypothesis.

So we can say: Accidents history does affect car prices

## Does fuel type affect prices?

Null Hypothesis: The mean prices are equal across all fuel types

Alternate Hypothesis: At least one fuel type has a different mean price

```python
data.fuel_type.value_counts()
```

```python
gasoline = data[data.fuel_type == 'Gasoline']['price']
hybrid = data[data.fuel_type == 'Hybrid']['price']
e85 = data[data.fuel_type == 'E85 Flex Fuel']['price']
diesel = data[data.fuel_type == 'Diesel']['price']
plug_in = data[data.fuel_type == 'Plug-In Hybrid']['price']
```

```python
t_stat, p_value = stats.f_oneway(
                    gasoline,
                    hybrid,
                    e85,
                    diesel,
                    plug_in
)

p_value
```

Since P-Value is less than 0.05. So we will reject the null hypothesis.

So we can say: At least one fuel type has different mean price

## Is mileage correlated with price?

Null Hypothesis: There is no linear correlation

Alternate Hypothesis: There is a significant linear correlation

```python
data[['milage', 'price']].corr()
```

```python
r_coeff, p_value = stats.pearsonr(data['milage'], data['price'])
r_coeff, p_value
```

Pearson correlation analysis showed a statistically significant negative correlation between milage and price. This indicates that vehicles with higher mileage tend to have lower prices

## Is horsepower significantly correlated with vehicle price?

Null Hypothesis: There is no linear correlation between horsepower and vehicle price

Alternative Hypothesis: There is a significant linear correlation between horsepower and vehicle price

```python
data.HP.isnull().sum()
```

```python
hp_data = data.dropna(subset=['HP'])

r_coeff, p_value = stats.pearsonr(hp_data['HP'], hp_data['price'])
r_coeff, p_value
```

Since P-Value < 0.05, we will reject null hypothesis.

Pearson correlation analysis relvealed a statistically significant moderate positive correlation between horsepower and vehicle price. This suggests that vehicles with higher horsepower tend to have higher prices
